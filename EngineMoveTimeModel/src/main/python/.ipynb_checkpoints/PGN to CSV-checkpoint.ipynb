```python
import chess.pgn

# Do note that the clock times are currently in centiseconds, not seconds.
pgn = open('../resources/lichess_db_univ_2019-01 (edited).pgn')

list_of_games = []

for x in range(0,100000): # can change how many games to load. There are millions of games in this .pgn
    list_of_games.append(chess.pgn.read_game(pgn))
```

```python
set_of_variants = set()

for game in list_of_games:
    set_of_variants.add(game.headers['Event'])

set_of_variants
# the different "types" of game played in the dataset.
# we'll want to exclude anything other than regular chess.
```

```python
cleaned_list_of_games = []

for game in list_of_games:
    has_clock_times = True # We'll filter out games where there isn't clock time data for every move
    number_of_moves = 0 # We'll filter out games where there aren't at least 4 moves
    termination_type = game.headers['Termination'] # We'll filter for games which ended normally (checkmate or loss on time)
    started_and_ended_normally = termination_type == 'Normal' or termination_type == 'Time forfeit'
    variant = game.headers['Event'] # We'll filter for rated games as they're more serious and hopefully have less nonsensical time usage.
    standard_chess_variant = variant == 'Rated Blitz game' or variant == 'Rated Rapid game' or variant == 'Rated Bullet game' or variant == 'Rated Classical game'
    
    for childNode in game.mainline():
        number_of_moves += 1
        if (childNode.clock() is None):
            has_clock_times = False
            
    if has_clock_times and number_of_moves >= 4 and started_and_ended_normally and standard_chess_variant:
        cleaned_list_of_games.append(game)
```

```python
print('Percentage of games that made it through filtering:', f"{len(cleaned_list_of_games)/len(list_of_games):.2%}")
```

```python
array_of_forfeits = []
for game in cleaned_list_of_games:
    if game.headers['Termination'] == 'Time forfeit':
        array_of_forfeits.append(game)

print('Percentage of games that ended on time:', f"{len(array_of_forfeits)/len(cleaned_list_of_games):.2%}")
```

```python
import numpy as np

# helper function for node_to_data to make the (12, 8, 8) of booleans feature representing the pieces
def childNode_to_pieceArrays(childNode):
    pieceArrays = []
    for colour in [True, False]:  # True is white, False is black
        for pieceType in range(1, 7): # 1 is pawn, 2 is knight, 3 is bishop, 4 is rook, 5 is queen, 6 is king
            pieceArrays.append(np.reshape(childNode.board().pieces(pieceType, colour).tolist(), (8,8)))
    return np.stack(pieceArrays)  # shape: (12, 8, 8)
    
# helper function for game_to_row
def node_to_data(node):
    clock = node.clock() / 100 # to convert centiseconds to seconds
    board = node.board()
    number_of_legal_moves = board.legal_moves.count()
    in_check = board.is_check()
    pieces = childNode_to_pieceArrays(node)
    tuple = (pieces, clock, number_of_legal_moves, in_check)
    return tuple

# helper function for game_to_row
def game_to_time_control_data(game):
    time_control = game.headers['TimeControl'].split('+')
    initial_time = int(time_control[0])
    increment = int(time_control[1])
    return (initial_time, increment)

# Turns a game into a list of tuples. 
# The tuples represent the features (including target feature of how much time was spent on the position)
def game_to_rows(game):
    initial_time, increment = game_to_time_control_data(game)
    temp_list = []
    for childNode in game.mainline():
        temp_list.append(node_to_data(childNode))

    list_of_rows = []
    for x in range(2, len(temp_list) - 1):
        pieces, opponent_time_remaining, number_of_legal_moves, in_check = temp_list[x]
        time_remaining_for_this_position = temp_list[x-1][1]
        time_after_this_position = temp_list[x+1][1]
        n_minus_2_clock = temp_list[x-2][1]

        # max(num, 0) to deal with negative move times (can occur when someone pre-moves and should obviously be interpreted as 0 seconds)
        # Round to the hundredths place because that's the level of precision the dataset actually has... 
        # ...this eliminates floating-point error
        time_opponent_took = round(max(n_minus_2_clock - opponent_time_remaining + increment, 0), 2)
        time_taken_for_this_position = round(max(time_remaining_for_this_position - time_after_this_position + increment, 0), 2)

        row_tuple = (pieces, initial_time, increment, time_remaining_for_this_position, time_opponent_took, number_of_legal_moves, in_check, time_taken_for_this_position)
        list_of_rows.append(row_tuple)

    return list_of_rows
```

```python
list_for_dataframe = []
for game in cleaned_list_of_games:
    list_for_dataframe += game_to_rows(game)

print('Length of list for dataframe', len(list_for_dataframe))
```
